```python
import os
import json
import yaml
import pickle
import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping
from prophet import Prophet

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Hospital Bed Occupancy Forecasting"
DATA_PATH = os.path.join(BASE_DIR, "hospital_directory.csv")

MODEL_DIR = os.path.join(BASE_DIR, "models")
CONFIG_DIR = os.path.join(BASE_DIR, "configs")
RESULT_DIR = os.path.join(BASE_DIR, "results")
SCALER_DIR = os.path.join(BASE_DIR, "scalers")

for d in [MODEL_DIR, CONFIG_DIR, RESULT_DIR, SCALER_DIR]:
    os.makedirs(d, exist_ok=True)

# ============================================================
# LOAD DATA
# ============================================================

df = pd.read_csv(DATA_PATH, low_memory=False)

# Clean numeric column
df["Total_Num_Beds"] = pd.to_numeric(df["Total_Num_Beds"], errors="coerce")
df = df.dropna(subset=["Total_Num_Beds"])

# ============================================================
# AGGREGATE BEDS (SIMULATED OCCUPANCY BASE)
# ============================================================

daily_base = int(df["Total_Num_Beds"].sum() * 0.7)  # assume 70% avg occupancy

days = 365
dates = pd.date_range(start="2023-01-01", periods=days, freq="D")

# Simulate realistic hospital load
np.random.seed(42)
occupancy = daily_base + np.random.randint(-200, 200, size=days)

ts_df = pd.DataFrame({
    "date": dates,
    "occupied_beds": occupancy
})

# ============================================================
# SAVE CONFIG
# ============================================================

config = {
    "dataset_type": "Hospital Directory",
    "date_strategy": "Synthetic Daily Timeline",
    "target": "occupied_beds (simulated)",
    "models": ["Prophet", "LSTM", "Hybrid"],
    "forecast_horizon_days": 30
}

with open(os.path.join(CONFIG_DIR, "config.yaml"), "w") as f:
    yaml.dump(config, f)

# ============================================================
# PROPHET MODEL
# ============================================================

prophet_df = ts_df.rename(columns={"date": "ds", "occupied_beds": "y"})

prophet = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=True
)

prophet.fit(prophet_df)

future = prophet.make_future_dataframe(periods=30)
prophet_forecast = prophet.predict(future)

prophet_preds = prophet_forecast.tail(30)[["ds", "yhat"]]

with open(os.path.join(MODEL_DIR, "prophet_model.pkl"), "wb") as f:
    pickle.dump(prophet, f)

# ============================================================
# LSTM MODEL
# ============================================================

scaler = MinMaxScaler()
scaled = scaler.fit_transform(ts_df[["occupied_beds"]])

with open(os.path.join(SCALER_DIR, "scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

def create_sequences(data, seq_len=14):
    X, y = [], []
    for i in range(len(data) - seq_len):
        X.append(data[i:i+seq_len])
        y.append(data[i+seq_len])
    return np.array(X), np.array(y)

SEQ_LEN = 14
X, y = create_sequences(scaled, SEQ_LEN)

split = int(0.8 * len(X))
X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

lstm = Sequential([
    LSTM(64, input_shape=(SEQ_LEN, 1)),
    Dense(1)
])

lstm.compile(optimizer="adam", loss="mse")

lstm.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=40,
    batch_size=16,
    callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
    verbose=1
)

lstm.save(os.path.join(MODEL_DIR, "lstm_model.h5"))

# ============================================================
# LSTM FUTURE FORECAST
# ============================================================

last_seq = scaled[-SEQ_LEN:]
lstm_preds = []

for _ in range(30):
    pred = lstm.predict(last_seq.reshape(1, SEQ_LEN, 1), verbose=0)
    lstm_preds.append(pred[0][0])
    last_seq = np.vstack([last_seq[1:], pred])

lstm_preds = scaler.inverse_transform(
    np.array(lstm_preds).reshape(-1, 1)
).flatten()

# ============================================================
# HYBRID FORECAST
# ============================================================

hybrid_preds = (prophet_preds["yhat"].values + lstm_preds) / 2

results = []

for d, p, l, h in zip(
    prophet_preds["ds"].astype(str),
    prophet_preds["yhat"],
    lstm_preds,
    hybrid_preds
):
    results.append({
        "date": d,
        "prophet_prediction": float(p),
        "lstm_prediction": float(l),
        "hybrid_prediction": float(h)
    })

with open(os.path.join(RESULT_DIR, "hybrid_forecast.json"), "w") as f:
    json.dump(results, f, indent=4)

print("✅ Hospital Bed Occupancy Forecasting Completed Successfully")
```
