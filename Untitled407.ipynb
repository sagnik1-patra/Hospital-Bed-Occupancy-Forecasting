```python
import os
import json
import yaml
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

from prophet import Prophet

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Hospital Bed Occupancy Forecasting"
DATA_PATH = os.path.join(BASE_DIR, "hospital_directory.csv")

MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULT_DIR = os.path.join(BASE_DIR, "results")
GRAPH_DIR = os.path.join(BASE_DIR, "graphs")

for d in [MODEL_DIR, RESULT_DIR, GRAPH_DIR]:
    os.makedirs(d, exist_ok=True)

# ============================================================
# LOAD & PREP DATA
# ============================================================

df = pd.read_csv(DATA_PATH, low_memory=False)
df["Total_Num_Beds"] = pd.to_numeric(df["Total_Num_Beds"], errors="coerce")
df = df.dropna(subset=["Total_Num_Beds"])

# Synthetic time series
days = 365
dates = pd.date_range("2023-01-01", periods=days, freq="D")
base = int(df["Total_Num_Beds"].sum() * 0.7)

np.random.seed(42)
occupied = base + np.random.randint(-200, 200, size=days)

ts_df = pd.DataFrame({
    "date": dates,
    "occupied_beds": occupied
})

# ============================================================
# PROPHET MODEL
# ============================================================

prophet_df = ts_df.rename(columns={"date": "ds", "occupied_beds": "y"})
prophet = Prophet(yearly_seasonality=True, weekly_seasonality=True)
prophet.fit(prophet_df)

future = prophet.make_future_dataframe(periods=30)
prophet_forecast = prophet.predict(future)

prophet_preds = prophet_forecast.tail(30)["yhat"].values

with open(os.path.join(MODEL_DIR, "prophet_model.pkl"), "wb") as f:
    pickle.dump(prophet, f)

# ============================================================
# LSTM MODEL
# ============================================================

scaler = MinMaxScaler()
scaled = scaler.fit_transform(ts_df[["occupied_beds"]])

def create_seq(data, n=14):
    X, y = [], []
    for i in range(len(data)-n):
        X.append(data[i:i+n])
        y.append(data[i+n])
    return np.array(X), np.array(y)

X, y = create_seq(scaled)
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

lstm = Sequential([
    LSTM(64, input_shape=(X.shape[1], 1)),
    Dense(1)
])

lstm.compile(optimizer="adam", loss="mse")

lstm.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=40,
    batch_size=16,
    callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
    verbose=1
)

lstm.save(os.path.join(MODEL_DIR, "lstm_model.h5"))

# ============================================================
# LSTM FUTURE FORECAST
# ============================================================

last_seq = scaled[-14:]
lstm_preds = []

for _ in range(30):
    pred = lstm.predict(last_seq.reshape(1,14,1), verbose=0)
    lstm_preds.append(pred[0][0])
    last_seq = np.vstack([last_seq[1:], pred])

lstm_preds = scaler.inverse_transform(
    np.array(lstm_preds).reshape(-1,1)
).flatten()

# ============================================================
# HYBRID
# ============================================================

hybrid_preds = (prophet_preds + lstm_preds) / 2
actual = ts_df["occupied_beds"].tail(30).values
dates_30 = ts_df["date"].tail(30).astype(str).values

# ============================================================
# METRICS
# ============================================================

metrics = {
    "MAE": mean_absolute_error(actual, hybrid_preds),
    "RMSE": np.sqrt(mean_squared_error(actual, hybrid_preds)),
    "R2": r2_score(actual, hybrid_preds)
}

# ============================================================
# SAVE CSV & JSON
# ============================================================

result_df = pd.DataFrame({
    "date": dates_30,
    "actual": actual,
    "prophet": prophet_preds,
    "lstm": lstm_preds,
    "hybrid": hybrid_preds
})

result_df.to_csv(os.path.join(RESULT_DIR, "results.csv"), index=False)

with open(os.path.join(RESULT_DIR, "predictions.json"), "w") as f:
    json.dump(result_df.to_dict(orient="records"), f, indent=4)

# ============================================================
# GRAPHS
# ============================================================

# Accuracy Graph
plt.figure()
plt.bar(metrics.keys(), metrics.values())
plt.title("Model Accuracy Metrics")
plt.savefig(os.path.join(GRAPH_DIR, "accuracy_graph.png"))
plt.show()

# Comparison Graph
plt.figure()
plt.plot(dates_30, prophet_preds, label="Prophet")
plt.plot(dates_30, lstm_preds, label="LSTM")
plt.plot(dates_30, hybrid_preds, label="Hybrid", linewidth=3)
plt.xticks(rotation=45)
plt.legend()
plt.title("Model Comparison")
plt.savefig(os.path.join(GRAPH_DIR, "comparison_graph.png"))
plt.show()

# Result Graph
plt.figure()
plt.plot(dates_30, actual, label="Actual")
plt.plot(dates_30, hybrid_preds, label="Hybrid Prediction")
plt.legend()
plt.title("Actual vs Predicted")
plt.savefig(os.path.join(GRAPH_DIR, "result_graph.png"))
plt.show()

# Prediction Graph
plt.figure()
plt.plot(dates_30, hybrid_preds, color="green")
plt.title("30-Day Bed Occupancy Forecast")
plt.savefig(os.path.join(GRAPH_DIR, "prediction_graph.png"))
plt.show()

# Heatmap
plt.figure()
sns.heatmap(result_df[["actual","prophet","lstm","hybrid"]].corr(), annot=True)
plt.title("Correlation Heatmap")
plt.savefig(os.path.join(GRAPH_DIR, "correlation_heatmap.png"))
plt.show()

print("✅ ALL RESULTS, GRAPHS, CSV & JSON GENERATED SUCCESSFULLY")
```
