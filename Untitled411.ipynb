```python
import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Hospital Bed Occupancy Forecasting"
DATA_PATH = os.path.join(BASE_DIR, "hospital_directory.csv")

MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULT_DIR = os.path.join(BASE_DIR, "results")
GRAPH_DIR = os.path.join(BASE_DIR, "graphs")

for d in [MODEL_DIR, RESULT_DIR, GRAPH_DIR]:
    os.makedirs(d, exist_ok=True)

# ============================================================
# DATA PREPARATION (SYNTHETIC TIME SERIES)
# ============================================================

df = pd.read_csv(DATA_PATH, low_memory=False)
df["Total_Num_Beds"] = pd.to_numeric(df["Total_Num_Beds"], errors="coerce")
df = df.dropna(subset=["Total_Num_Beds"])

days = 365
dates = pd.date_range("2023-01-01", periods=days, freq="D")

base = int(df["Total_Num_Beds"].sum() * 0.7)
np.random.seed(42)
occupied = base + np.random.randint(-200, 200, size=days)

ts_df = pd.DataFrame({
    "date": dates,
    "occupied_beds": occupied
})

# ============================================================
# SCALING
# ============================================================

scaler = MinMaxScaler()
scaled = scaler.fit_transform(ts_df[["occupied_beds"]])

with open(os.path.join(MODEL_DIR, "bis_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

# ============================================================
# SEQUENCE CREATION
# ============================================================

def create_sequences(data, n=14):
    X, y = [], []
    for i in range(len(data) - n):
        X.append(data[i:i+n])
        y.append(data[i+n])
    return np.array(X), np.array(y)

SEQ_LEN = 14
X, y = create_sequences(scaled, SEQ_LEN)
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ============================================================
# MODEL BUILDER
# ============================================================

def build_lstm(units):
    model = Sequential([
        LSTM(int(units), input_shape=(SEQ_LEN, 1)),
        Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model

def fitness(units):
    model = build_lstm(units)
    model.fit(X_train, y_train, epochs=8, batch_size=16, verbose=0)
    preds = model.predict(X_test, verbose=0)
    return mean_squared_error(y_test, preds)

# ============================================================
# BAT ALGORITHM (GLOBAL SEARCH)
# ============================================================

bats = np.random.randint(32, 96, size=6)
frequencies = np.random.rand(6)
loudness = np.ones(6)
pulse_rate = np.random.rand(6)

fitness_scores = [fitness(b) for b in bats]
best_bat = bats[np.argmin(fitness_scores)]

for _ in range(5):
    for i in range(len(bats)):
        freq = frequencies[i]
        bats[i] = bats[i] + freq * (bats[i] - best_bat)
        bats[i] = int(np.clip(bats[i], 32, 128))

        score = fitness(bats[i])
        if score < fitness_scores[i]:
            fitness_scores[i] = score
        else:
            bats[i] = best_bat

    best_bat = bats[np.argmin(fitness_scores)]

print(f"🔍 BA Best Units: {best_bat}")

# ============================================================
# AIS PHASE (CLONAL SELECTION)
# ============================================================

ais_best = int(best_bat)

for _ in range(5):
    clones = [
        ais_best + np.random.randint(-6, 6)
        for _ in range(4)
    ]
    clones = [int(np.clip(c, 32, 128)) for c in clones]

    clone_scores = [fitness(c) for c in clones]
    best_clone = clones[np.argmin(clone_scores)]

    if fitness(best_clone) < fitness(ais_best):
        ais_best = best_clone

best_units = ais_best
print(f"✅ Optimized Units via BA+AIS: {best_units}")

# ============================================================
# FINAL BIS MODEL
# ============================================================

bis_model = build_lstm(best_units)

bis_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=40,
    batch_size=16,
    callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
    verbose=1
)

bis_model.save(os.path.join(MODEL_DIR, "bis_lstm_model.h5"))

# ============================================================
# FUTURE PREDICTION
# ============================================================

last_seq = scaled[-SEQ_LEN:]
bis_preds = []

for _ in range(30):
    pred = bis_model.predict(last_seq.reshape(1,SEQ_LEN,1), verbose=0)
    bis_preds.append(pred[0][0])
    last_seq = np.vstack([last_seq[1:], pred])

bis_preds = scaler.inverse_transform(
    np.array(bis_preds).reshape(-1,1)
).flatten()

actual = ts_df["occupied_beds"].tail(30).values
dates_30 = ts_df["date"].tail(30).astype(str).values

# ============================================================
# METRICS
# ============================================================

metrics = {
    "MAE": mean_absolute_error(actual, bis_preds),
    "RMSE": np.sqrt(mean_squared_error(actual, bis_preds)),
    "R2": r2_score(actual, bis_preds)
}

# ============================================================
# SAVE CSV & JSON
# ============================================================

result_df = pd.DataFrame({
    "date": dates_30,
    "actual": actual,
    "bis_prediction": bis_preds
})

result_df.to_csv(os.path.join(RESULT_DIR, "bis_results.csv"), index=False)

with open(os.path.join(RESULT_DIR, "bis_predictions.json"), "w") as f:
    json.dump(result_df.to_dict(orient="records"), f, indent=4)

# ============================================================
# GRAPHS (bis_ PREFIX)
# ============================================================

# Accuracy Graph
plt.figure()
plt.bar(metrics.keys(), metrics.values())
plt.title("BIS Accuracy Metrics")
plt.savefig(os.path.join(GRAPH_DIR, "bis_accuracy_graph.png"))
plt.show()

# Result Graph
plt.figure()
plt.plot(dates_30, actual, label="Actual")
plt.plot(dates_30, bis_preds, label="BIS Prediction")
plt.legend()
plt.title("BIS Actual vs Prediction")
plt.savefig(os.path.join(GRAPH_DIR, "bis_result_graph.png"))
plt.show()

# Prediction Graph
plt.figure()
plt.plot(dates_30, bis_preds, color="green")
plt.title("BIS 30-Day Forecast")
plt.savefig(os.path.join(GRAPH_DIR, "bis_prediction_graph.png"))
plt.show()

# Heatmap
plt.figure()
sns.heatmap(result_df[["actual","bis_prediction"]].corr(), annot=True)
plt.title("BIS Correlation Heatmap")
plt.savefig(os.path.join(GRAPH_DIR, "bis_heatmap.png"))
plt.show()

print("✅ BA + AIS (BIS) PIPELINE COMPLETED SUCCESSFULLY")
```
