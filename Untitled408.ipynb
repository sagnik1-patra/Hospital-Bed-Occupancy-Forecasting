```python
import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Hospital Bed Occupancy Forecasting"
DATA_PATH = os.path.join(BASE_DIR, "hospital_directory.csv")

MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULT_DIR = os.path.join(BASE_DIR, "results")
GRAPH_DIR = os.path.join(BASE_DIR, "graphs")

for d in [MODEL_DIR, RESULT_DIR, GRAPH_DIR]:
    os.makedirs(d, exist_ok=True)

# ============================================================
# DATA PREPARATION (SYNTHETIC TIME SERIES)
# ============================================================

df = pd.read_csv(DATA_PATH, low_memory=False)
df["Total_Num_Beds"] = pd.to_numeric(df["Total_Num_Beds"], errors="coerce")
df = df.dropna(subset=["Total_Num_Beds"])

days = 365
dates = pd.date_range("2023-01-01", periods=days, freq="D")
base = int(df["Total_Num_Beds"].sum() * 0.7)

np.random.seed(42)
occupied = base + np.random.randint(-200, 200, size=days)

ts_df = pd.DataFrame({
    "date": dates,
    "occupied_beds": occupied
})

# ============================================================
# SCALING
# ============================================================

scaler = MinMaxScaler()
scaled = scaler.fit_transform(ts_df[["occupied_beds"]])

with open(os.path.join(MODEL_DIR, "hybrid_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

# ============================================================
# SEQUENCE CREATION
# ============================================================

def create_seq(data, n=14):
    X, y = [], []
    for i in range(len(data) - n):
        X.append(data[i:i+n])
        y.append(data[i+n])
    return np.array(X), np.array(y)

SEQ_LEN = 14
X, y = create_seq(scaled, SEQ_LEN)
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ============================================================
# AIS + CSA OPTIMIZATION (SIMPLIFIED & STABLE)
# ============================================================

def build_lstm(units):
    model = Sequential([
        LSTM(int(units), input_shape=(SEQ_LEN, 1)),
        Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model

def fitness(units):
    model = build_lstm(units)
    model.fit(
        X_train, y_train,
        epochs=10,
        batch_size=16,
        verbose=0
    )
    preds = model.predict(X_test, verbose=0)
    return mean_squared_error(y_test, preds)

# --- AIS Phase ---
ais_population = np.random.randint(32, 96, size=6)
ais_scores = [fitness(u) for u in ais_population]
best_ais_units = ais_population[np.argmin(ais_scores)]

# --- CSA Phase ---
csa_memory = best_ais_units
for _ in range(5):
    candidate = csa_memory + np.random.randint(-8, 8)
    candidate = np.clip(candidate, 32, 128)
    if fitness(candidate) < fitness(csa_memory):
        csa_memory = candidate

best_units = int(csa_memory)

print(f"✅ Optimized LSTM Units via Hybrid AIS+CSA: {best_units}")

# ============================================================
# FINAL HYBRID MODEL
# ============================================================

hybrid_lstm = build_lstm(best_units)

hybrid_lstm.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=40,
    batch_size=16,
    callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
    verbose=1
)

hybrid_lstm.save(os.path.join(MODEL_DIR, "hybrid_lstm_model.h5"))

# ============================================================
# PREDICTION
# ============================================================

last_seq = scaled[-SEQ_LEN:]
hybrid_preds = []

for _ in range(30):
    pred = hybrid_lstm.predict(last_seq.reshape(1,SEQ_LEN,1), verbose=0)
    hybrid_preds.append(pred[0][0])
    last_seq = np.vstack([last_seq[1:], pred])

hybrid_preds = scaler.inverse_transform(
    np.array(hybrid_preds).reshape(-1,1)
).flatten()

actual = ts_df["occupied_beds"].tail(30).values
dates_30 = ts_df["date"].tail(30).astype(str).values

# ============================================================
# METRICS
# ============================================================

metrics = {
    "MAE": mean_absolute_error(actual, hybrid_preds),
    "RMSE": np.sqrt(mean_squared_error(actual, hybrid_preds)),
    "R2": r2_score(actual, hybrid_preds)
}

# ============================================================
# SAVE CSV & JSON
# ============================================================

result_df = pd.DataFrame({
    "date": dates_30,
    "actual": actual,
    "hybrid_prediction": hybrid_preds
})

result_df.to_csv(
    os.path.join(RESULT_DIR, "hybrid_results.csv"), index=False
)

with open(os.path.join(RESULT_DIR, "hybrid_predictions.json"), "w") as f:
    json.dump(result_df.to_dict(orient="records"), f, indent=4)

# ============================================================
# GRAPHS (ALL WITH hybrid_ PREFIX)
# ============================================================

# Accuracy Graph
plt.figure()
plt.bar(metrics.keys(), metrics.values())
plt.title("Hybrid AIS+CSA Accuracy")
plt.savefig(os.path.join(GRAPH_DIR, "hybrid_accuracy_graph.png"))
plt.show()

# Result Graph
plt.figure()
plt.plot(dates_30, actual, label="Actual")
plt.plot(dates_30, hybrid_preds, label="Hybrid Prediction")
plt.legend()
plt.title("Hybrid Actual vs Prediction")
plt.savefig(os.path.join(GRAPH_DIR, "hybrid_result_graph.png"))
plt.show()

# Prediction Graph
plt.figure()
plt.plot(dates_30, hybrid_preds, color="green")
plt.title("Hybrid 30-Day Forecast")
plt.savefig(os.path.join(GRAPH_DIR, "hybrid_prediction_graph.png"))
plt.show()

# Heatmap
plt.figure()
sns.heatmap(result_df[["actual","hybrid_prediction"]].corr(), annot=True)
plt.title("Hybrid Correlation Heatmap")
plt.savefig(os.path.join(GRAPH_DIR, "hybrid_heatmap.png"))
plt.show()

print("✅ HYBRID AIS + CSA PIPELINE COMPLETED SUCCESSFULLY")
```
