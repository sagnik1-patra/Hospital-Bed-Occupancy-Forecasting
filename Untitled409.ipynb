```python
import os
import json
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ============================================================
# PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Hospital Bed Occupancy Forecasting"
DATA_PATH = os.path.join(BASE_DIR, "hospital_directory.csv")

MODEL_DIR = os.path.join(BASE_DIR, "models")
RESULT_DIR = os.path.join(BASE_DIR, "results")
GRAPH_DIR = os.path.join(BASE_DIR, "graphs")

for d in [MODEL_DIR, RESULT_DIR, GRAPH_DIR]:
    os.makedirs(d, exist_ok=True)

# ============================================================
# DATA PREPARATION (SYNTHETIC TIME SERIES)
# ============================================================

df = pd.read_csv(DATA_PATH, low_memory=False)
df["Total_Num_Beds"] = pd.to_numeric(df["Total_Num_Beds"], errors="coerce")
df = df.dropna(subset=["Total_Num_Beds"])

days = 365
dates = pd.date_range("2023-01-01", periods=days, freq="D")

base = int(df["Total_Num_Beds"].sum() * 0.7)
np.random.seed(42)
occupied = base + np.random.randint(-200, 200, size=days)

ts_df = pd.DataFrame({
    "date": dates,
    "occupied_beds": occupied
})

# ============================================================
# SCALING
# ============================================================

scaler = MinMaxScaler()
scaled = scaler.fit_transform(ts_df[["occupied_beds"]])

with open(os.path.join(MODEL_DIR, "pis_scaler.pkl"), "wb") as f:
    pickle.dump(scaler, f)

# ============================================================
# SEQUENCE CREATION
# ============================================================

def create_sequences(data, n=14):
    X, y = [], []
    for i in range(len(data) - n):
        X.append(data[i:i+n])
        y.append(data[i+n])
    return np.array(X), np.array(y)

SEQ_LEN = 14
X, y = create_sequences(scaled, SEQ_LEN)
split = int(0.8 * len(X))

X_train, X_test = X[:split], X[split:]
y_train, y_test = y[:split], y[split:]

# ============================================================
# MODEL BUILDER
# ============================================================

def build_lstm(units):
    model = Sequential([
        LSTM(int(units), input_shape=(SEQ_LEN, 1)),
        Dense(1)
    ])
    model.compile(optimizer="adam", loss="mse")
    return model

def fitness(units):
    model = build_lstm(units)
    model.fit(X_train, y_train, epochs=8, batch_size=16, verbose=0)
    preds = model.predict(X_test, verbose=0)
    return mean_squared_error(y_test, preds)

# ============================================================
# AIS PHASE (GLOBAL SEARCH)
# ============================================================

ais_population = np.random.randint(32, 96, size=6)
ais_scores = [fitness(u) for u in ais_population]

best_ais_units = ais_population[np.argmin(ais_scores)]

print(f"🔍 AIS Best Units: {best_ais_units}")

# ============================================================
# PSO PHASE (LOCAL REFINEMENT)
# ============================================================

particles = best_ais_units + np.random.randint(-10, 10, size=5)
velocities = np.zeros_like(particles)

pbest = particles.copy()
pbest_scores = [fitness(p) for p in particles]

gbest = pbest[np.argmin(pbest_scores)]

for _ in range(5):
    for i in range(len(particles)):
        velocities[i] = (
            0.5 * velocities[i]
            + 0.8 * (pbest[i] - particles[i])
            + 0.9 * (gbest - particles[i])
        )
        particles[i] += velocities[i]
        particles[i] = int(np.clip(particles[i], 32, 128))

        score = fitness(particles[i])
        if score < pbest_scores[i]:
            pbest[i] = particles[i]
            pbest_scores[i] = score

    gbest = pbest[np.argmin(pbest_scores)]

best_units = int(gbest)
print(f"✅ Optimized Units via AIS+PSO: {best_units}")

# ============================================================
# FINAL PIS MODEL
# ============================================================

pis_model = build_lstm(best_units)

pis_model.fit(
    X_train, y_train,
    validation_data=(X_test, y_test),
    epochs=40,
    batch_size=16,
    callbacks=[EarlyStopping(patience=5, restore_best_weights=True)],
    verbose=1
)

pis_model.save(os.path.join(MODEL_DIR, "pis_lstm_model.h5"))

# ============================================================
# FUTURE PREDICTION
# ============================================================

last_seq = scaled[-SEQ_LEN:]
pis_preds = []

for _ in range(30):
    pred = pis_model.predict(last_seq.reshape(1,SEQ_LEN,1), verbose=0)
    pis_preds.append(pred[0][0])
    last_seq = np.vstack([last_seq[1:], pred])

pis_preds = scaler.inverse_transform(
    np.array(pis_preds).reshape(-1,1)
).flatten()

actual = ts_df["occupied_beds"].tail(30).values
dates_30 = ts_df["date"].tail(30).astype(str).values

# ============================================================
# METRICS
# ============================================================

metrics = {
    "MAE": mean_absolute_error(actual, pis_preds),
    "RMSE": np.sqrt(mean_squared_error(actual, pis_preds)),
    "R2": r2_score(actual, pis_preds)
}

# ============================================================
# SAVE CSV & JSON
# ============================================================

result_df = pd.DataFrame({
    "date": dates_30,
    "actual": actual,
    "pis_prediction": pis_preds
})

result_df.to_csv(os.path.join(RESULT_DIR, "pis_results.csv"), index=False)

with open(os.path.join(RESULT_DIR, "pis_predictions.json"), "w") as f:
    json.dump(result_df.to_dict(orient="records"), f, indent=4)

# ============================================================
# GRAPHS
# ============================================================

# Accuracy Graph
plt.figure()
plt.bar(metrics.keys(), metrics.values())
plt.title("PIS Accuracy Metrics")
plt.savefig(os.path.join(GRAPH_DIR, "pis_accuracy_graph.png"))
plt.show()

# Result Graph
plt.figure()
plt.plot(dates_30, actual, label="Actual")
plt.plot(dates_30, pis_preds, label="PIS Prediction")
plt.legend()
plt.title("PIS Actual vs Prediction")
plt.savefig(os.path.join(GRAPH_DIR, "pis_result_graph.png"))
plt.show()

# Prediction Graph
plt.figure()
plt.plot(dates_30, pis_preds, color="green")
plt.title("PIS 30-Day Forecast")
plt.savefig(os.path.join(GRAPH_DIR, "pis_prediction_graph.png"))
plt.show()

# Heatmap
plt.figure()
sns.heatmap(result_df[["actual","pis_prediction"]].corr(), annot=True)
plt.title("PIS Correlation Heatmap")
plt.savefig(os.path.join(GRAPH_DIR, "pis_heatmap.png"))
plt.show()

print("✅ AIS + PSO (PIS) PIPELINE COMPLETED SUCCESSFULLY")
```
